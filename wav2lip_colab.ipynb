```python
# 2. Tải ảnh avatar và audio từ Google Drive
import gdown

# Ảnh avatar của Hỉ
gdown.download('https://drive.google.com/uc?export=download&id=1bF-TMCWJ8TH4uIs38aerL4yRpMYgQl5k', 'avatar.jpg', quiet=False)

# Audio đã tạo từ Edge TTS
gdown.download('https://drive.google.com/uc?export=download&id=1fmo0GSiexm6Un26E0rhErP30ad_P0MBu', 'audio.mp3', quiet=False)

print('✅ Đã tải ảnh và audio')
```

```python
# 3. Chạy Wav2Lip
!python inference.py \
    --checkpoint_path wav2lip_gan.pth \
    --face avatar.jpg \
    --audio audio.mp3 \
    --outfile results.mp4 \
    --resize_factor 2
print('✅ Video đã tạo!')
```

```python
# 4. Hiển thị video
from IPython.display import HTML
from base64 import b64encode

video = open('results.mp4', 'rb').read()
data_url = 'data:video/mp4;base64,' + b64encode(video).decode()
HTML(f'<video width=480 controls><source src="{data_url}" type="video/mp4"></video>')
```
